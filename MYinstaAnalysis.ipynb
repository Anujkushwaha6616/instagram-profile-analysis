```python
with open("finaldata 6.txt", encoding="utf-8") as fl:
    data = fl.read()
```

```python
chunks = data.split("\n\n")
chunks = [c for c in chunks if len(c)>3]
# chunk = chunks[0].split("\n")
# chunk
# print("User name:",chunk[0])
# no_of_posts = int(chunk[1].split(" post")[0].replace(",",""))
# print("Number of post:",no_of_posts)
# print("number of followers:",chunk[2])
```

```python
def parse_chunk(chunk):
    chunk = chunk.strip()
    sep_chunk = chunk.split("\n")
    user_name = sep_chunk[0]
    no_of_post = int(sep_chunk[1].split(" post")[0].replace(",",""))
    no_of_followers = float(sep_chunk[2].split(" follower")[0].replace(",","").replace("K","").replace("M",""))
    if "K" in sep_chunk[2]:
        no_of_followers = int(no_of_followers * 1000)
    elif "M" in sep_chunk[2]:
        no_of_followers = int(no_of_followers * 100000)
    else:
        no_of_followers = int(no_of_followers)
        
    no_of_following = sep_chunk[3].split(" following")[0].replace(",","").replace("K","").replace("M","")
    if "K" in sep_chunk[3]:
        no_of_following = int(no_of_following * 1000)
    elif "M" in sep_chunk[3]:
        no_of_following = int(no_of_following * 1000000)
    else:
        no_of_following = int(no_of_following)

    name = sep_chunk[4]
    if len(sep_chunk)>5:
        type_of_page = sep_chunk[5]
        bio = "\n".join(sep_chunk[6:])
    else:
        type_of_page = "Unknown"
        bio = ""

    return {"user_name":user_name,"no_of_post":no_of_post,"no_of_followers":no_of_followers,"no_of_following":no_of_following,"name":name,"type_of_page":type_of_page}
    # print(user_name,"     ",no_of_post,"     ",no_of_followers,"       ",no_of_following,"     ",name,"         " ,type_of_page)


    
```

```python
all_chunks = []
for chunk in chunks:
    all_chunks.append(parse_chunk(chunk))
```

```python
max = 0
max_post_chunk = 0
for cc in all_chunks:
    if max < cc['no_of_post']:
        max = cc['no_of_post']
        max_post_chunk = cc
print(f"{max_post_chunk['user_name']} = has maximum post total post are : {max}")
```

```python
max = 0
for cc in all_chunks:
    if max < cc['no_of_followers']:
        max = cc['no_of_followers']
        max_no_of_followers = cc
print(f"{max_no_of_followers['user_name']} = has maximum followers : {max}")
```
